### Wyznaczyć i zinterpretować:
- miary centralności dla wierzchołków,
- składowe spójne oraz k-spójność wierzchołkowa/krawędziowa,
- klikę maksymalną (maximal clique), klikę największą (maximum clique) oraz wskazać ich rząd.

#### Wyjaśnienia
- betweeness – odsetek wszystkich najkrótszych ścieżek, które przechodzą przez dany wierzchołek. Pokazują co jest potencjalnym mostem (zakładając ze zalezy nam na najkrótszej trasie).
- efficiency – średnia odwrotności odległości od danego węzła do wszystkich pozostałych.
- pagerank – bierze pod uwagę istotność połączenia, tj. połączenie z jednym duzym wierzchołkiem moze dać większą wartość niz z wieloma małymi.

- składowe spójne – podgrafy, w których kazdy da się przejść do kazdego wierzchołka
- k-spójność wierzchołkowa / krawędziowa – minimalna liczba wierzchołków / krawędzi, których usunięcie spowoduje niespójność,
- klika - podgraf gdzie kazdy ma połączenie z kazdym,
- klika maksymalna – taka klika do której nie da się juz dołozyć wierzchołka (dowolnego), bo rozspójni klikę,
- klika największa – największa maksymalna klika,
- rząd kliki – z ilu wierzchołków się składa

```python
N = len(graph)
betweeness = nx.betweenness_centrality(graph)
efficiency = {u: val / (N - 1) for u, val in nx.harmonic_centrality(graph).items()}
page_rank = nx.pagerank(graph)

components = list(nx.connected_components(graph))
num_components = len(components)
largest_cc = max(components, key=len)
largest_subgraph = graph.subgraph(largest_cc).copy()
node_conn_sub = nx.node_connectivity(largest_subgraph)
edge_conn_sub = nx.edge_connectivity(largest_subgraph)
all_cliques_maximal = list(nx.find_cliques(graph))
maximum_clique = max(all_cliques_maximal, key=len)


print(f"Liczba spójnych składowych: {num_components}")
print(f"Rozmiar największej składowej: {len(largest_cc)} wierzchołków (z {len(graph)})")
print(f"k-spójność wierzchołkowa: {node_conn_sub}")
print(f"k-spójność krawędziowa: {edge_conn_sub}")



print(f"Liczba wszystkich klik maksymalnych w grafie: {len(all_cliques_maximal)}")
print(f"Rząd największej kliki: {len(maximum_clique)}")
print(f"Wierzchołki kliki największej: {[graph.nodes[n]['name'] for n in maximum_clique]}")
```

### Dla każdej miary centralności:
- przedstawić pięć wierzchołków o najwyższych wartościach,
- określić, jaką rolę mogą pełnić te wierzchołki.

Betweeness - usunięcie tych wierzchołków najbardziej wydłuzy średnio najkrótsze trasę, wracając do historyjki z brakiem internetu, dana uczelnia musiałaby (średnio) skontaktować się z większą liczbą "znajomych znajomych", aby dotrzeć do docelowej uczelni.

Efficiency - z tych wierzchołków najszybciej dostaniemy się do wszystkich innych (zapominając o niespójności), wskazówka gdzie zacząć, chcąc współpracować ze wszystkimi (lub szukając po omacku)


PageRank - zakładając ze chcemy nawiązać współpracę z wielką/znaną uczelnią i nie mozemy się do niej dostać bezpośrednio (bo nie pracuje juz z malo znanymi organizacjami), mozemy próbować dostać się do nich przez uczelnię z wysokim wskaznikiem pagerank.

```python
measures = {
    "Betweeness": betweeness,
    "Efficiency": efficiency,
    "PageRank": page_rank,
}
for name, d in measures.items():
    print(f"{name}")
    top5 = sorted(d.items(), key=lambda x: x[1], reverse=True)[:5]
    for i, (node, val) in enumerate(top5, 1):
        print(f"{i}. {graph.nodes[node]['name']}: {val}, (stopień: {graph.degree[node]})")
    print()
```
